```python
import pandas as pd
import plotly.express as px
```

```python
# remove unnecessary columns
df = df.drop(columns=["Last_Updated","Android_Ver"])
df.head()
```

```python
df.shape
```

```python
nan_rows = df[df.Rating.isna()]
print(nan_rows.shape)
```

```python
# remove NaN based on the  rating column
df_apps_clean = df.dropna(subset=['Rating'])
df_apps_clean.head()
```

```python
# identify duplicates and remove them
duplicated_rows = df_apps_clean[df_apps_clean.duplicated()]
print(duplicated_rows.shape)
duplicated_rows.head()
```

```python
df_apps_clean[df_apps_clean.App=='Instagram']
```

```python
df_apps_clean=df_apps_clean.drop_duplicates(subset=['App','Type','Price'])
```

```python
df_apps_clean.shape
```

```python
# apps highly rated
highly_rated_apps = df_apps_clean.loc[df_apps_clean.Rating.idxmax()]
highly_rated_apps
```

```python
# apps with highest reviews
reviewed_high_apps = df_apps_clean.sort_values(by=["Reviews"],ascending=False)
reviewed_high_apps
```

```python
rated_high_apps = df_apps_clean.sort_values(by=["Rating"],ascending=False)
rated_high_apps
```

```python
# visualize using plotly
content_rating = df_apps_clean["Content_Rating"].value_counts()
content_rating
```

```python
# content_rating visualization using pie
fig = px.pie(labels=content_rating.index,
             values=content_rating.values,
             title="Content Rating",
             names=content_rating.index
             )
fig.update_traces(textposition='outside',textinfo='percent+label')
fig.show()
```

```python
# content_rating visualization using donut
fig = px.pie(labels=content_rating.index,
             values=content_rating.values,
             title="Content Rating",
             names=content_rating.index,
             hole=0.6
             )
fig.update_traces(textposition='outside',textinfo='percent',textfont_size=15)
fig.show()
```

```python
df_apps_clean.Category.nunique()
```

```python
top_10_category = df_apps_clean.Category.value_counts()[:10]
top_10_category
```

```python
bar = px.bar(x=top_10_category.index,
             y=top_10_category.values)
bar.show()
```

```python
category_installs = df_apps_clean.groupby('Category').agg({'Installs': pd.Series.sum})
category_installs.sort_values('Installs', ascending=True, inplace=True)
```

```python
h_bar = px.bar(x = category_installs.Installs,
               y = category_installs.index,
               orientation='h',
               title='Category Popularity')

h_bar.update_layout(xaxis_title='Number of Downloads', yaxis_title='Category')
h_bar.show()
```

```python
category_concentration = df_apps_clean.groupby('Category').agg({'App': pd.Series.count,'Installs': pd.Series.sum})
category_concentration.sort_values("Installs",ascending=False,inplace=True)
```

```python
scatter = px.scatter(category_concentration, # data
                    x='App', # column name
                    y='Installs',
                    title='Category Concentration',
                    size='App',
                    hover_name=category_concentration.index,
                    color='Installs')

scatter.update_layout(xaxis_title="Number of Apps (Lower=More Concentrated)",
                      yaxis_title="Installs",
                      yaxis=dict(type='log'))

scatter.show()
```

```python
# nested columns
len(df_apps_clean.Genres.unique())
```

```python
df_apps_clean.Genres.value_counts(ascending=False).head(10)
```

```python
stack = df_apps_clean.Genres.str.split(';', expand=True).stack()
print(f'We now have a single column with shape: {stack.shape}')
num_genres = stack.value_counts()
print(f'Number of genres: {len(num_genres)}')
```

```python
num_genres
```

```python
bar = px.bar(x = num_genres.index[:15], # index = category name
             y = num_genres.values[:15], # count
             title='Top Genres',
             hover_name=num_genres.index[:15],
             color=num_genres.values[:15],
             color_continuous_scale='Agsunset')

bar.update_layout(xaxis_title='Genre',yaxis_title='Number of Apps',coloraxis_showscale=False)

bar.show()
```

```python
df_free_vs_paid = df_apps_clean.groupby(["Category", "Type"], as_index=False).agg({'App': pd.Series.count})
df_free_vs_paid.head()
```

```python
g_bar = px.bar(df_free_vs_paid,
               x='Category',
               y='App',
               title='Free vs Paid Apps by Category',
               color='Type',
               barmode='group')

g_bar.update_layout(xaxis_title='Category',
                    yaxis_title='Number of Apps',
                    xaxis={'categoryorder':'total descending'},
                    yaxis=dict(type='log'))

g_bar.show()
```
